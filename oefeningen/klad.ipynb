## 1. Een lijst

Maak een **lijst** met vijf omschrijvingen van uitgaven, als tekst.

````python
ls = ["Supermarkt","Slager","Bakker","Bart Smit", "Blokker"]

print(ls)
````

---
## 2. Een dictionary

Maak een **dictionary** die elke omschrijving aan een bedrag koppelt.

````python
UitgavenDict = {

    "Supermarkt": 2.75,
    "Slager": 22,
    "Bakker" : 45,
    "Bart Smit" : 45 , 
    "Blokker": 2.75
}


print(UitgavenDict)
````

## 4. Een bedrag via de dictionary

Print het bedrag van één omschrijving, door de dictionary te bevragen.

````python
bedragSlager = UitgavenDict["Slager"]


print(bedragSlager)
````

## 5. Welk type is het?


Print van je lijst en van je dictionary het type met `type()`.

Print er ook één bedrag en één omschrijving bij. Vier keer `type()` dus, en kijk
of je alle vier kunt benoemen.

````python
print(type(UitgavenDict))
print(type(ls))


print(type(UitgavenDict["Bakker"]))
print(type(ls[1]))
````

## 7. Alles optellen

Schrijf een `for`-lus die alle bedragen bij elkaar optelt, en print het totaal.

*Hint:* loop je door de lijst of door de dictionary? Bij een dictionary levert `for` je
standaard de **sleutels**, niet de bedragen. Zoekterm voor de waarden: `.values()`

Let op de dubbele punt en de indentatie.

````python
totaal_getal = 0

for x in UitgavenDict:
    totaal_getal += UitgavenDict[x]


print(totaal_getal)
````

---
## 8. Een voorwaarde

Schrijf een `if` die print of het totaal boven de 50 euro ligt of niet.

````python
if totaal_getal > 50:
    print("Het bedrag ligt boven de 50 euro")
else:
    print("Het bedrag ligt onder de 50 euro")
````
